We can check the correlation with the target for the columns with 5-30% missing data. #TODO Explain p-value 0.05

```python
from scipy import stats
import numpy as np

# Merge x_train with target column from y_train using 'id'
df_temp = df.merge(y_train[['Id', '_MICHD']], on='Id', how='inner')

# Identify columns with >5% missing
missing_pct = df.isna().mean() * 100
cols_missing_over_5 = missing_pct[missing_pct>5].index

# Compute p-values for correlation with the target
pvalues = {}
for col in cols_missing_over_5:
    # Remove NaN values for both column and target
    valid_mask = df_temp[col].notna() & df_temp['_MICHD'].notna()
    if valid_mask.sum() > 0:
        corr, pval = stats.pearsonr(df_temp.loc[valid_mask, col],
                                    df_temp.loc[valid_mask, '_MICHD'])
        pvalues[col] = pval

pvalues = pd.Series(pvalues).sort_values(ascending=False)
pvalues
```

Next we can remove a couple more columns that are highly correlated to reduce redundancy, #TODO Justify 80%

```python
missing_pct = df.isna().mean() * 100
cols_missing_over_5 = missing_pct[missing_pct > 5].index.tolist()
df_subset = df.copy()[cols_missing_over_5].dropna()
corr_matrix = df_subset.corr()

# Find highly correlated pairs (correlation > 0.8)
print("\nHighly correlated pairs (|correlation| > 0.8):")
high_corr_pairs = []
for i in range(len(corr_matrix.columns)):
    for j in range(i+1, len(corr_matrix.columns)):
        if abs(corr_matrix.iloc[i, j]) > 0.8:
            high_corr_pairs.append({
                'Feature 1': corr_matrix.columns[i],
                'Feature 2': corr_matrix.columns[j],
                'Correlation': corr_matrix.iloc[i, j]
            })

if high_corr_pairs:
    high_corr_df = pd.DataFrame(high_corr_pairs)
    print(high_corr_df.to_string(index=False))
else:
    print("No pairs found with |correlation| > 0.8")
```

Now that we've reduced the number of columns with missing data, interpolate missing data based on skew, high skew data use media, low skew use mean.

```python
# Interpolate missing values based on skewness recommendations
from scipy.stats import skew

# Get columns with missing values
cols_with_missing = df.columns[df.isnull().any()].tolist()

# Separate columns by interpolation method based on skewness
median_cols = []
mean_cols = []

for col in cols_with_missing:
    skewness = skew(df[col].dropna())
    if abs(skewness) > 0.5:
        median_cols.append(col)
    else:
        mean_cols.append(col)

# Fill with median for skewed columns (correct way)
for col in median_cols:
    df[col] = df[col].fillna(df[col].median())

# Fill with mean for normally distributed columns (correct way)
for col in mean_cols:
    df[col] = df[col].fillna(df[col].mean())

print(f"Filled {len(median_cols)} columns with median")
print(f"Filled {len(mean_cols)} columns with mean")
```
